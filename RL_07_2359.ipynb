```python
n = int(input("Enter number of time steps: "))

rewards = []

for i in range(n):
    reward = float(input(f"Enter reward at time step {i}: "))
    rewards.append(reward)

gamma = float(input("Enter discount factor (gamma): "))

undiscounted_return = sum(rewards)

discounted_return = 0

for t in range(n):
    discounted_return += (gamma ** t) * rewards[t]

print("\nEntered rewards:", rewards)
print("Undiscounted Return:", undiscounted_return)
print("Discounted Return:", discounted_return)
```

```python
ROWS = 2
COLS = 3

START = (0, 0)
GOAL = (1, 2)
OBSTACLE = (0, 1)

GAMMA = 0.9

ACTIONS = {
    "up": (-1, 0),
    "down": (1, 0),
    "left": (0, -1),
    "right": (0, 1)
}

STATES = [
    (r, c)
    for r in range(ROWS)
    for c in range(COLS)
    if (r, c) != OBSTACLE
]


def reward_function(state):
    if state == GOAL:
        return 10
    return -1


def transition_function(state, action):
    dr, dc = ACTIONS[action]

    next_state = (
        state[0] + dr,
        state[1] + dc
    )

    r, c = next_state

    if (
        r < 0 or r >= ROWS
        or c < 0 or c >= COLS
        or next_state == OBSTACLE
    ):
        next_state = state

    return next_state
```

```python
print("State Space:", STATES)
print("Action Space:", list(ACTIONS.keys()))
print("Discount Factor:", GAMMA)

state = START

print("\nStarting State:", state)

while state != GOAL:
    action = input(
        "Enter action (up/down/left/right): "
    ).lower()

    if action not in ACTIONS:
        print("Invalid action")
        continue

    next_state = transition_function(
        state, action
    )

    reward = reward_function(next_state)

    print("Current State:", state)
    print("Action:", action)
    print("Next State:", next_state)
    print("Reward:", reward)

    state = next_state

print("\nGoal Reached!")
```
