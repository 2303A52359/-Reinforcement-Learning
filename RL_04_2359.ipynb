```python
import math

machines = ['A', 'B', 'C']

M = 3

N_play = 10

N = [0, 0, 0]
R = [0, 0, 0]
Q = [0, 0, 0]
UCB = [0, 0, 0]

total_reward = 0


for play in range(1, N_play + 1):


    if play <= M:
        selected = play - 1
    else:

        for i in range(M):
            UCB[i] = Q[i] + math.sqrt((2 * math.log(play)) / N[i])

        selected = UCB.index(max(UCB))

    print("\nPlay", play)
    print("Selected Machine:", machines[selected])

    reward = float(input("Enter reward: "))

    N[selected] += 1

    R[selected] += reward
    total_reward += reward

    Q[selected] = Q[selected] + (reward - Q[selected]) / N[selected]

    print("N =", N)
    print("R =", R)
    print("Q =", [round(x, 2) for x in Q])


print("\n========== FINAL RESULTS ==========")

print("\nMachine\tN\tR\tUCB\tQ")


for i in range(M):
    if N[i] > 0:
        UCB[i] = Q[i] + math.sqrt((2 * math.log(N_play)) / N[i])
    else:
        UCB[i] = float('inf')

for i in range(M):
    print(machines[i], "\t", N[i], "\t", R[i],
          "\t", round(UCB[i], 2),
          "\t", round(Q[i], 2))

print("\nTotal Reward =", total_reward)
```
